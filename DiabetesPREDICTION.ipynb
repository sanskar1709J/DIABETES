```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, classification_report
import joblib

# 1. Load the Massive CDC BRFSS Dataset (250,000+ Real Patient Records)
url = "https://raw.githubusercontent.com/Helmy2/Diabetes-Health-Indicators/main/diabetes_binary_health_indicators_BRFSS2015.csv"
print("Downloading and loading CDC dataset...")
df = pd.read_csv(url)

# 2. Inspect data shape and clean duplicates
print(f"Original dataset shape: {df.shape}")
df = df.drop_duplicates()
print(f"Shape after removing duplicates: {df.shape}")

# 3. Define Features (X) and Target (y)
# Target: Diabetes_binary (0 = No Diabetes, 1 = Prediabetes or Diabetes)
X = df.drop('Diabetes_binary', axis=1)
y = df['Diabetes_binary']

# 4. Train/Test Split (Stratified to maintain class balance)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 5. Scale features for optimal numerical performance
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 6. Train an Optimized XGBoost Classifier on 250k+ Records
print("Training advanced XGBoost model on massive data...")
xgb_model = XGBClassifier(
    n_estimators=150,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='logloss'
)
xgb_model.fit(X_train_scaled, y_train)

# 7. Evaluate Model Performance
preds_proba = xgb_model.predict_proba(X_test_scaled)[:, 1]
roc_auc = roc_auc_score(y_test, preds_proba)
print(f"🔥 New Massive-Data XGBoost ROC-AUC Score: {roc_auc:.4f}")

# 8. Save the upgraded model, scaler, and exact feature list using joblib
joblib.dump(xgb_model, 'diabetes_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(X.columns.tolist(), 'feature_names.pkl')
print("✅ Large-scale model, scaler, and feature list successfully saved!")
```
